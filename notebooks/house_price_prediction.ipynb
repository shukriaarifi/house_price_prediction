```python
df=df[df["price"]>0]
df.shape
```

49 rows with zero price are removed

```python
df=df[(df["bedrooms"]>0) & (df["bathrooms"]>0)]
df.shape
```

2 rows with zero bathrooms and bedrooms are removed

```python
q1=df["price"].quantile(0.25)
q3=df["price"].quantile(0.75)
IQR=q3-q1
lower_bound=q1-1.5*IQR
upper_bound=q3+1.5*IQR

print("Q1:",q1)
print("Q3:",q3)
print("IQR:",IQR)
print("Lower bound:",lower_bound)
print("Upper bound:",upper_bound)
```

```python
outliers=(df[(df["price"]<lower_bound) | (df["price"]>upper_bound)]).copy()
print("Number of outliers:", len(outliers))
```

```python
outliers[["price", "sqft_living", "sqft_lot", "bedrooms", "bathrooms", "floors", "waterfront", "condition"]].sort_values("price", ascending=False).head(20)
```

prices outliers were investigated using the IQR method

```python
outliers["price_per_sqft"]=(outliers["price"]/outliers["sqft_living"])
outliers[["price","sqft_living","price_per_sqft"]].sort_values("price_per_sqft",ascending=False).head(10)
```

```python
features=["bedrooms", "bathrooms", "sqft_living", "sqft_lot", "floors", "waterfront", "view", "condition", "sqft_above", "sqft_basement", "yr_built", "yr_renovated"]
```

```python
y=df["price"]
x=df[features]
```

selected target and features

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test=train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train.head()
```

```python
y_train.head()
```

```python
x_train.shape
```

```python
y_train.shape
```

splited data into train teast 80 20

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
model.fit(x_train, y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
y_pred=model.predict(x_test)
```

model was trained using training data and used to predict house prices on the test set

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
import numpy as np
from sklearn.metrics import r2_score
```

```python
mae=mean_absolute_error(y_test, y_pred)
mse=mean_squared_error(y_test, y_pred)
rmse=np.sqrt(mse)
r2=r2_score(y_test, y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R^2:", r2)
```

```python
plt.figure(figsize=(7, 5))
plt.scatter(y_test, y_pred)

plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], linestyle="--")

plt.xlabel("Actual price")
plt.ylabel("Predicted price")
plt.title("Actual vs predicted price")

plt.show()
```

```python
result=pd.DataFrame({"Actual price": y_test, "Predicted price": y_pred})
result["Error"]=result["Actual price"] - result["Predicted price"]

result.head(10)
```

```python
result["Absolute error"]=result["Error"].abs()
result.sort_values("Absolute error", ascending=False).head(10)
```
