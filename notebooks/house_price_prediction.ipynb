```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df["price"])
plt.xlabel("price")
plt.ylabel("num of houses")
plt.title("distrbution of price")
plt.show()
```

```python
plt.hist(df["price"]/1_000_000,bins=30)
plt.xlabel("price(Million $)")
plt.ylabel("num of houses")
plt.title("distrbution of price")
plt.show()
```

there are some large prices that should be limited for analysing details

```python
plt.hist(df["price"]/1_000_000,bins=30)
plt.xlim(0,5)
plt.xlabel("price(Million $)")
plt.ylabel("num of houses")
plt.title("distrbution of price(0-5 million $)")
plt.show()
```

```python
plt.scatter(df["sqft_living"],df["price"])
plt.xlabel("living area")
plt.ylabel("price")
plt.title("living area vs house price")
plt.show()
```
